```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# Dataset: CO2 Emissions by Country 2000-2022
# Source: Our World in Data (https://ourworldindata.org/co2-emissions)
df = pd.read_csv('../data/co2_emissions.csv')
print(f"Loaded: {len(df)} rows | Countries: {df['Country'].nunique()} | Years: {df['Year'].min()}-{df['Year'].max()}")
print(df.head())
```

```python
# Explore before building

print("Countries:", df['Country'].unique())
print("\nCO2 range:", df['CO2_Mt'].min(), "to", df['CO2_Mt'].max(), "Mt")
print("\nRegional averages (2022):")
print(df[df['Year']==2022].groupby('Region')['CO2_Mt'].mean().sort_values(ascending=False).round(1))
```

## Task 1 - A line chart showing CO2 per Capita for all countries in the dataset

**Requirements:**
- All countries shown (for context), but only **one highlighted in colour** — your choice which
- All other lines in grey (#EEEEEE), thinner
- Highlighted country **labelled directly** at the end of its line (not in a legend)
- Insight title that names the highlighted country and its story

```python
highlight_country = "China"
highlight_color = "#0072B2"

fig = go.Figure()

# Show every other country in faint grey
for country in df["Country"].unique():
    if country == highlight_country:
        continue

    country_data = df[df["Country"] == country].sort_values("Year")
    fig.add_trace(go.Scatter(
        x=country_data["Year"],
        y=country_data["CO2_per_capita"],
        mode="lines",
        line=dict(color="#EEEEEE", width=1),
        showlegend=False,
        hovertemplate=f"{country}<br>Year: %{{x}}<br>CO2 per capita: %{{y}}<extra></extra>",
    ))

# Draw China on top so it stands out
highlighted = (
    df[df["Country"] == highlight_country]
    .sort_values("Year")
)

fig.add_trace(go.Scatter(
    x=highlighted["Year"],
    y=highlighted["CO2_per_capita"],
    mode="lines",
    line=dict(color=highlight_color, width=3),
    showlegend=False,
    hovertemplate="China<br>Year: %{x}<br>CO2 per capita: %{y}<extra></extra>",
))

# Put the label beside China's final data point
fig.add_annotation(
    x=highlighted["Year"].iloc[-1],
    y=highlighted["CO2_per_capita"].iloc[-1],
    text=highlight_country,
    showarrow=False,
    xshift=8,
    xanchor="left",
    font=dict(color=highlight_color, size=13),
)

fig.update_layout(
    title="China's CO2 per capita climbed from 1.32 to 4.37, 2000-2022",
    xaxis_title="Year",
    yaxis_title="CO2 per capita",
    template="plotly_white",
    showlegend=False,
    margin=dict(r=100),
)

# Leave room after the final year for the direct label
fig.update_xaxes(
    range=[df["Year"].min(), df["Year"].max() + 3],
    showline=False,
    zeroline=False,
    gridcolor="#F2F2F2",
)
fig.update_yaxes(
    showline=False,
    zeroline=False,
    gridcolor="#F2F2F2",
)

fig.show()
```

## Task 2 — A line chart showing CO2 emissions over time for all Asian countries in the dataset, with one country highlighted.

**same requirements as Task1**

```python
# Task — Multi-series line with highlight
# YOUR CODE HERE
asia_df = df[df["Region"] == "Asia"]
highlight_country = "China"
highlight_color = "#0072B2"

fig = go.Figure()

# Draw all other Asian countries in grey
for country in asia_df["Country"].unique():
    if country == highlight_country:
        continue

    country_data = asia_df[asia_df["Country"] == country].sort_values("Year")
    fig.add_trace(go.Scatter(
        x=country_data["Year"],
        y=country_data["CO2_Mt"],
        mode="lines",
        line=dict(color="#EEEEEE", width=1),
        showlegend=False,
        hovertemplate=(
            f"{country}<br>Year: %{{x}}"
            "<br>CO2 emissions: %{y:,.0f} Mt<extra></extra>"
        ),
    ))

# Draw China in colour on top
highlighted = (
    asia_df[asia_df["Country"] == highlight_country]
    .sort_values("Year")
)

fig.add_trace(go.Scatter(
    x=highlighted["Year"],
    y=highlighted["CO2_Mt"],
    mode="lines",
    line=dict(color=highlight_color, width=3),
    showlegend=False,
    hovertemplate=(
        "China<br>Year: %{x}"
        "<br>CO2 emissions: %{y:,.0f} Mt<extra></extra>"
    ),
))

# Label China at the end of its line
fig.add_annotation(
    x=highlighted["Year"].iloc[-1],
    y=highlighted["CO2_Mt"].iloc[-1],
    text=highlight_country,
    showarrow=False,
    xshift=8,
    xanchor="left",
    font=dict(color=highlight_color, size=13),
)

fig.update_layout(
    title="China's CO2 emissions more than tripled from 2000 to 2022",
    xaxis_title="Year",
    yaxis_title="CO2 emissions (Mt)",
    template="plotly_white",
    showlegend=False,
    margin=dict(r=100),
)

fig.update_xaxes(
    range=[asia_df["Year"].min(), asia_df["Year"].max() + 3],
    showline=False,
    zeroline=False,
    gridcolor="#F2F2F2",
)
fig.update_yaxes(
    showline=False,
    zeroline=False,
    gridcolor="#F2F2F2",
)

fig.show()
```
