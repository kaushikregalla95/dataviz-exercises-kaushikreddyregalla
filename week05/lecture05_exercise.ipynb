## Distribution Charts: Airbnb London

**Rules:**
1. Cap price outliers at 95th percentile — annotate this
2. Every chart has a **median/mean reference line** with annotation
3. Insight title names the distribution shape or key finding
4. Colour has meaning — don't use colour just for decoration

---


```python
import pandas as pd
import plotly.express as px
import numpy as np

# Dataset: Airbnb London Listings

df = pd.read_csv('../data/airbnb_london.csv')
print(f"Loaded: {len(df)} listings")
print(df.describe().round(1))
```

```python
p95 = df['price'].quantile(0.95)
df_cap = df[df['price'] <= p95]
print(f"95th percentile price: £{p95:.0f}")
print(df_cap.groupby('room_type')['price'].describe().round(1))
```

## Task 1 — Histogram: price by room type (overlapping distributions)

**What to build:** A histogram showing price distributions for **Entire home/apt vs Private room** (exclude Shared room — too few observations) overlaid on the same chart.

**Requirements:**
- Both room types on the same chart (use `color='room_type'`)
- `barmode='overlay'` with `opacity=0.6` so both distributions are visible
- A vertical line for the median of EACH room type, differently coloured
- Insight title comparing the two distributions

```python
# Task 1
# YOUR CODE HERE

# Task 1
t1 = df_cap[df_cap['room_type'].isin(['Entire home/apt', 'Private room'])]

colors = {'Entire home/apt': '#2E75B6', 'Private room': '#70AD47'}
medians = t1.groupby('room_type')['price'].median()
ratio = medians['Entire home/apt'] / medians['Private room']

fig = px.histogram(
    t1, x='price', color='room_type',
    barmode='overlay', opacity=0.6, nbins=50,
    color_discrete_map=colors,
    labels={'price': 'Nightly Price (£)', 'room_type': 'Room type'},
    height=500
)

# Median line for each room type, in its own colour
positions = {'Entire home/apt': 'top right', 'Private room': 'top left'}
for room, m in medians.items():
    fig.add_vline(x=m, line_dash='dash', line_color=colors[room], line_width=2,
                  annotation_text=f'<b>{room} median: £{m:.0f}</b>',
                  annotation_font_color=colors[room],
                  annotation_position=positions[room])

# Disclose the outlier cap
fig.add_annotation(xref='paper', yref='paper', x=1, y=-0.18, showarrow=False,
                   text=f'Prices capped at 95th percentile (£{p95:.0f})',
                   font=dict(size=11, color='grey'))

fig.update_layout(
    title=f'Entire homes cost about {ratio:.1f}x more than private rooms, and their prices are far more spread out',
    plot_bgcolor='white', paper_bgcolor='white',
    yaxis_title='Number of listings',
    margin=dict(b=80)
)
fig.show()
```

## Task 2 — Box plot: listing activity by borough

**What to build:** A **horizontal box plot** comparing listing activity (reviews per month) across London boroughs — reviews per month is a proxy for how frequently a listing is booked.

**Requirements:**
- Horizontal orientation (borough names are long)
- Sorted by median reviews per month (most active at top)
- Highlight the **two most active** boroughs in a different colour
- Outliers shown as individual points
- Insight title naming the two busiest boroughs

> 💡 Some listings have zero reviews — these are new or inactive listings. Filter them out with before plotting

```python
# Task 2
# YOUR CODE HERE

# Task 2
t2 = df[df['reviews_per_month'] > 0]          # remove new/inactive listings

# Sort boroughs by median (ascending, so the most active ends up at the top)
order = t2.groupby('neighbourhood')['reviews_per_month'].median().sort_values().index.tolist()
top2 = order[-2:][::-1]                        # two most active boroughs

# Red for the top two, grey for the rest
color_map = {b: ('#E63946' if b in top2 else '#BBBBBB') for b in order}

fig = px.box(
    t2, x='reviews_per_month', y='neighbourhood',
    orientation='h',
    color='neighbourhood', color_discrete_map=color_map,
    category_orders={'neighbourhood': order},
    points='outliers',
    labels={'reviews_per_month': 'Reviews per month', 'neighbourhood': ''},
    height=900
)

# Overall median reference line
overall_med = t2['reviews_per_month'].median()
fig.add_vline(x=overall_med, line_dash='dash', line_color='black',
              annotation_text=f'London median: {overall_med:.1f}',
              annotation_position='top')

fig.update_layout(
    title=f'{top2[0]} and {top2[1]} are the busiest boroughs, with the highest median reviews per month',
    plot_bgcolor='white', paper_bgcolor='white',
    showlegend=False
)
fig.show()
```
